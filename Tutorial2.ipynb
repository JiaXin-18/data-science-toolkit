```python
from IPython.core.interactiveshell import no_op
#Part 1-Google Colab and Python
#1. Create a new notebook and code the following

#1. Create a numpy array of size 2 x 2

import numpy as np
array_2x2 = np.array([
    [1,2],
    [3,4]
])

print("This is my 2x2 array")
print(array_2x2)
```

```python
#2. Create a random numpy array of size 3 x 4
# The first number is row and second number is column
#Thus allow you to make  an array (3,4)

random_3x4 = np.random.rand(3,4)

print("Random3x4 array")
print(random_3x4)
```

```python
#3. Create a numpy array of ones and a numpy array of zeroes
ones_array = np.ones((3,3))
zeros_array = np.zeros((3,3))

print ("Array of ones:")
print(ones_array)

print("Array of zeros")
print(zeros_array)
```

```python
#4. Create a random numpy array of size 6x3 and transpose it
random_6x3 = np.random.rand(6,3)
transposed_random_6x3 = np.transpose(random_6x3)

print("Random 6x3 array")
print(random_6x3)

print("Transposed random 6x3 array")
print(transposed_random_6x3)
```

```python
#5. Create two random numpy array of size n x p and p x m and then compute the dot product

first_array = np.random.rand(n, p)

second_array = np.random.rand(p, m)

result = np.dot(first_array, second_array)

print("First array:")
print(first_array)

print("Second array:")
print(second_array)

print("Dot product:")
print(result)
```
