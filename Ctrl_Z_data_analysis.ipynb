Count total transactions

```python
total = con.execute(f"""
    SELECT COUNT(*)
    FROM read_csv_auto('{file_path}')
""").fetchone()[0]

print("Total transactions:", total)
```

Analyse numerical data

A. USD_AMOUNT

```python
amount_stats = con.execute(f"""
    SELECT
        COUNT(USD_AMOUNT) AS total,
        AVG(USD_AMOUNT) AS mean,
        MEDIAN(USD_AMOUNT) AS median,
        MIN(USD_AMOUNT) AS minimum,
        MAX(USD_AMOUNT) AS maximum,
        STDDEV(USD_AMOUNT) AS standard_deviation,
        SKEWNESS(USD_AMOUNT) AS skewness
    FROM read_csv_auto('{file_path}')
""").df()

display(amount_stats)
```

B. TOTAL_DURATION

```python
duration_stats = con.execute(f"""
    SELECT
        COUNT(TOTAL_DURATION) AS total,
        AVG(TOTAL_DURATION) AS mean,
        MEDIAN(TOTAL_DURATION) AS median,
        MIN(TOTAL_DURATION) AS minimum,
        MAX(TOTAL_DURATION) AS maximum,
        STDDEV(TOTAL_DURATION) AS standard_deviation,
        SKEWNESS(TOTAL_DURATION) AS skewness
    FROM read_csv_auto('{file_path}')
""").df()

display(duration_stats)
```

Check missing values

```python
missing = con.execute(f"""
    SELECT
        COUNT(*) AS total_rows,
        COUNT(*) - COUNT(USD_AMOUNT) AS missing_amount,
        COUNT(*) - COUNT(TOTAL_DURATION) AS missing_duration,
        COUNT(*) - COUNT(CURRENCY_CODE) AS missing_currency,
        COUNT(*) - COUNT(BIC8_PATH) AS missing_path,
        COUNT(*) - COUNT(ANOMALY_TYPE) AS missing_anomaly,
        COUNT(*) - COUNT(ORDERING_TOKENS) AS missing_ordering_tokens,
        COUNT(*) - COUNT(BENEFICIARY_TOKENS) AS missing_beneficiary_tokens

    FROM read_csv_auto('{file_path}')
""").df()

display(missing.T)
```

2A. Explore currency distribution

```python
currency_stats = con.execute(f"""
    SELECT
        CURRENCY_CODE,
        COUNT(*) AS transaction_count,
        ROUND(
            COUNT(*) * 100.0 / SUM(COUNT(*)) OVER (),
            2
        ) AS percentage
    FROM read_csv_auto('{file_path}')
    GROUP BY CURRENCY_CODE
    ORDER BY transaction_count DESC
""").df()

display(currency_stats)
```

2B. Explore transaction types

```python
transaction_stats = con.execute(f"""
    SELECT
        TRANSACTION_TYPE_CODE,
        COUNT(*) AS transaction_count
    FROM read_csv_auto('{file_path}')
    GROUP BY TRANSACTION_TYPE_CODE
    ORDER BY transaction_count DESC
""").df()

display(transaction_stats)
```

2C. Explore customer types

```python
customer_stats = con.execute(f"""
    SELECT
        ORDERING_TYPE,
        BENEFICIARY_TYPE,
        COUNT(*) AS transaction_count
    FROM read_csv_auto('{file_path}')
    GROUP BY ORDERING_TYPE, BENEFICIARY_TYPE
    ORDER BY transaction_count DESC
""").df()

display(customer_stats)
```

2D. Number of participants

```python
participant_stats = con.execute(f"""
    SELECT
        PARTICIPANT_COUNT,
        COUNT(*) AS transaction_count,
        ROUND(AVG(TOTAL_DURATION), 2) AS avg_duration,
        MEDIAN(TOTAL_DURATION) AS median_duration
    FROM read_csv_auto('{file_path}')
    GROUP BY PARTICIPANT_COUNT
    ORDER BY PARTICIPANT_COUNT
""").df()

display(participant_stats)
```

2E. Compare transaction types and duration

```python
type_duration = con.execute(f"""
    SELECT
        TRANSACTION_TYPE_CODE,
        COUNT(*) AS transaction_count,
        ROUND(AVG(TOTAL_DURATION), 2) AS avg_duration,
        MEDIAN(TOTAL_DURATION) AS median_duration
    FROM read_csv_auto('{file_path}')
    GROUP BY TRANSACTION_TYPE_CODE
    ORDER BY avg_duration DESC
""").df()

display(type_duration)
```

2F. Participant count by transaction type

```python
participant_by_type = con.execute(f"""
    SELECT
        TRANSACTION_TYPE_CODE,
        PARTICIPANT_COUNT,
        COUNT(*) AS transaction_count,
        ROUND(AVG(TOTAL_DURATION), 2) AS avg_duration,
        MEDIAN(TOTAL_DURATION) AS median_duration
    FROM read_csv_auto('{file_path}')
    GROUP BY
        TRANSACTION_TYPE_CODE,
        PARTICIPANT_COUNT
    ORDER BY
        TRANSACTION_TYPE_CODE,
        PARTICIPANT_COUNT
""").df()

display(participant_by_type)
```

2G. Currency conversion vs duration

```python
conversion_stats = con.execute(f"""
    SELECT
        CURRENCY_CONVERSION_COUNT,
        COUNT(*) AS transaction_count,
        ROUND(AVG(TOTAL_DURATION), 2) AS avg_duration,
        MEDIAN(TOTAL_DURATION) AS median_duration
    FROM read_csv_auto('{file_path}')
    GROUP BY CURRENCY_CONVERSION_COUNT
    ORDER BY CURRENCY_CONVERSION_COUNT
""").df()

display(conversion_stats)
```

Participant count AND currency conversion count

```python
combined_stats = con.execute(f"""
    SELECT
        PARTICIPANT_COUNT,
        CURRENCY_CONVERSION_COUNT,
        COUNT(*) AS transaction_count,
        ROUND(AVG(TOTAL_DURATION), 2) AS avg_duration,
        MEDIAN(TOTAL_DURATION) AS median_duration
    FROM read_csv_auto('{file_path}')
    GROUP BY
        PARTICIPANT_COUNT,
        CURRENCY_CONVERSION_COUNT
    HAVING COUNT(*) >= 100
    ORDER BY
        PARTICIPANT_COUNT,
        CURRENCY_CONVERSION_COUNT
""").df()

display(combined_stats)
```

Whether transaction type explains some of the participant-count relationship

```python
type_participant_stats = con.execute(f"""
    SELECT
        TRANSACTION_TYPE_CODE,
        PARTICIPANT_COUNT,
        COUNT(*) AS transaction_count,
        ROUND(AVG(TOTAL_DURATION), 2) AS avg_duration,
        MEDIAN(TOTAL_DURATION) AS median_duration
    FROM read_csv_auto('{file_path}')
    GROUP BY
        TRANSACTION_TYPE_CODE,
        PARTICIPANT_COUNT
    HAVING COUNT(*) >= 100
    ORDER BY
        TRANSACTION_TYPE_CODE,
        PARTICIPANT_COUNT
""").df()

display(type_participant_stats)
```

Do transactions with gaps or placeholders in their routes take longer than transactions with continuous routes?

```python
continuity_stats = con.execute(f"""
    SELECT
        CONTINUITY_QFLAG,
        COUNT(*) AS transaction_count,
        ROUND(AVG(TOTAL_DURATION), 2) AS avg_duration,
        MEDIAN(TOTAL_DURATION) AS median_duration
    FROM read_csv_auto('{file_path}')
    GROUP BY CONTINUITY_QFLAG
    ORDER BY CONTINUITY_QFLAG
""").df()

display(continuity_stats)
```
