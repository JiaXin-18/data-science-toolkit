```python
#2. Create a random numpy array of size 3 x 4
# The first number is row and second number is column
#Thus allow you to make  an array (3,4)

random_3x4 = np.random.rand(3,4)

print("Random3x4 array")
print(random_3x4)
```

```python
#3. Create a numpy array of ones and a numpy array of zeroes
ones_array = np.ones((3,3))
zeros_array = np.zeros((3,3))

print ("Array of ones:")
print(ones_array)

print("Array of zeros")
print(zeros_array)
```
